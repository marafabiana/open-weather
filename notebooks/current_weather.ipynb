```python
import pandas as pd
from open_weather.services.temperatures import get_temperatures
from open_weather.visualization.pandas.Barchart import barchart

temperatures = get_temperatures("Brasilia,BR", "Lisbon,PT", "Stockholm,SE")

df = pd.DataFrame({"Temperature": temperatures})

barchart(
    df,
    title="Current temperature",
    x_label="City",
    y_label="Temperature (°C)"
)
```
