```python
import pandas as pd
from open_weather.services.temperatures import get_temperatures
from open_weather.visualization.pandas.Barchart import barchart

from open_weather.visualization.maps.bubble_map import bubble_map
from open_weather.services.weather import get_weather_data
temperatures = get_temperatures("Brasilia,BR", "Lisbon,PT", "Stockholm,SE")

df = pd.DataFrame({"Temperature": temperatures})
```

```python
barchart(
    df,
    title="Current temperature",
    x_label="City",
    y_label="Temperature (°C)"
)
```

```python
rows = get_weather_data("Stockholm,SE", "Oslo,NO", "Copenhagen,DK", "Madrid,ES", "Brasilia,BR")
df = pd.DataFrame(rows)

bubble_map(df, "Current temperature and pressure")
```
